# Series

> Series is an 1d array in Pandas, which is capable of holding any datatype (int ,float , python object)

>  **Syntax for creating series :**
  ```
  >> variable = pd.Series(collection)
  ```

> If you try to create series by using dictionary, then keys will be taken as index values and values of those respective keys will be considered as values of series

````python
a = pd.Series([10,20,30,40,60,50])
a
````

````python
a = pd.Series([10,20,30,40.55,60,50])
a
````

````python
a = pd.Series([10, 20, '30', 40, 50])
a
#In Pandas, string is also considered as object datatype.
````

EX:

      ```python

      ```

      It sn inbuild function which is used to check wheather the particular value is nan value or not.
      - It word fine for numpy array

      ```python
      np.isnan(value / array)
      ```

      It sn inbuild function which is used to check wheather the value  in Series or DataFrame is nan or not.
      
      Syntax:
      ```python
      pd.isnan(value / array / series / dataframe)
      ```

```txt
Question

Write a Python program using NumPy and Pandas to generate a synthetic student performance
dataset for 15 students.

The dataset should include student details, academic records, and final results.
Your program must follow the requirements below:

1. Generate the following fields:
      Student_Ids — Create IDs from 101 to 115.
      Name — Use a predefined list of 15 student names.
      Age — Randomly assign ages 13, 14, or 15 to each student.
      Class — Randomly assign class 8, 9, or 10.
      Gender — Randomly assign Male or Female.
      Subject — Randomly assign one subject from Maths, Science, Python, Java.
      Obtained Marks — Generate random marks between 300 and 400.
      Attendance — Generate random attendance between 75 and 100.

2. Calculate the following derived fields:
      Percentage = (Obtained Marks / 400) × 100
      Grade = 'A' if Percentage ≥ 90 else 'B'
      Final Result = 'Pass' if Grade is 'A' else 'Improvement required'

3. Introduce Missing Values:
      Randomly replace three entries each from:
          Name
          Subject
          Obtained Marks
      with NaN values.

4. Handle Missing Values in Derived Fields:
    If marks are NaN, then percentage, grade, and result should also be NaN.

5. Create a Pandas DataFrame containing all the above fields.
```

````python
import pandas as pd
import numpy as np
id =   np.arange(1,16)
name = ['uday','udayraj','pruthvi','pruthviraj','aishu','aishawarya','rani','machindra','arjun','janabai','ash','ram','mahadev','parvati','sham']
age = np.random.randint(13,16,15)
clas = np.random.randint(8,11,15)
gender = np.random.choice(['Male','Female'],15)
subject =  np.random.choice(['Maths', 'Science','Python','java'],15)
obtained_marks = np.random.randint(300,401,15)
attendance = np.random.randint(75,101,15)

df = pd.DataFrame({
    'student_Ids' : id,
    'name' : name,
    'age' : age,
    'class' : clas,
    'gender' : gender,
    'subject' : subject,
    'obtained_marks': obtained_marks,
    'attendance' : attendance
})
df
````

````python
# 2. Calculate the following derived fields:
#       Percentage = (Obtained Marks / 400) × 100
#       Grade = 'A' if Percentage ≥ 90 else 'B'
#       Final Result = 'Pass' if Grade is 'A' else 'Improvement required'

percentage = [round((marks/400)*100) for marks in df['obtained_marks']]
df['percentage'] = percentage
grade = ['A' if per>= 90 else 'B' for per in df['percentage']]
df['grades'] = grade
final_result = ['Pass' if grade == 'A' else 'Imporvement Required' for grade in df['grades']]
df['final_result'] = final_result


df
````

````python
# 3. Introduce Missing Values:
#       Randomly replace three entries each from:
#           Name
#           Subject
#           Obtained Marks
#       with NaN values.ddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddddx
for i in range(3):
  df.name[np.random.randint(1,15,1)] = np.nan
  df.subject[np.random.randint(1,15,1)]= np.nan
  df.obtained_marks[np.random.randint(1,15,1)]= np.nan
df
````

````python

# 4. Handle Missing Values in Derived Fields:
#     If marks are NaN, then percentage, grade, and result should also be NaN.

# count = 0
# for i in df['obtained_marks']:
#   if i.isnan():
#     df['Percentage'][count] = np.nan
#     df['Grade'][count] = np.nan
#     df['Final_result'][count] = np.nan
#   count += 1

percentage = [np.nan if pd.isna(i) else round((i / 400) * 100, 2) for i in obtained_marks]

grades = [np.nan if pd.isna(i) else ('A' if i >=90 else 'B') for i in percentage]

result = [np.nan if pd.isna(i) else ('Pass' if i == 'A' else 'Improvement required')
           for i in grades]
````

````python

# 5. Create a Pandas DataFrame containing all the above fields.
df = pd.DataFrame({
    'student_Ids' : id,
    'name' : name,
    'age' : age,
    'class' : clas,
    'gender' : gender,
    'subject' : subject,
    'obtained_marks': obtained_marks,
    'attendance' : attendance,
    'percentange' : percentage,
    'grades' : grades,
    'final_result' : result
})

df
````

# Inbuild Function Of DataFrame

1. head()

    It is an build function which will return top n row from the dataframe.

    Syntax:
    ```python
    df.head() ---> top 5 rows
        or
    df.head(n) ---> top n rows
    ```



3. info()

    It is an build function which will give complete information about the dataframe or dataset.

    Syntax:
    ```python
    df.info()
    ```

4. decribe()

    It is an build function which is use to give statistical information about dataframe.
    
    Syntax:
    ```python
    df.decribe() ---> work on numrical column
            OR
    df.describe(include = object) --->work on object columns
    ```

5. shape

    It is an inbuild attribute which will used to return number of row and  number of column in dataset in tuple formate.

    Syntax:
    ```python
    df.shape
    ```

6. columns

    It is an inbuild attribute which will used to give column names from the dataset.

    Syntax:
    ```python
    df.columns
    ```


7. unique()

    It is an inbuild function which is used to give unique value from the specific column.

    Syntax:
    ```python
    df.column_name.unique()
        or
    df['column_name'].unique()
    ```

    It will return an array of all unique value parsent inside a specific column.

    We can use unique() function for only one column at a time


8. sort_values()

    It is an inbuild function which will return sorted dataframe based on specific column

    Syntax:
    ```python
    df.sort_values(by = 'column_name') ---> sort in ascending order
                  or
    df.sort_values(by = 'column_name', ascending = False) ---> sort in descending order
    ```

9. value_counts():

    It is an inbuilt function which will returns the unique values with their respective counts.

    Syntax:
    ```
    df['column_name'].value_counts()
    ```

10. sample():

    It is an inbuilt function which will return random n rows from the dataframe.
      
    Syntax:
    ```
    df.sample(n)
    ```

## ***loc[]**

>It an inbuild function / indexer which is used to slice and access rows and columns using their labels(names).

>It select the data by row label (i.e. index) and column names.

> It support slicing for single column and multiple columns using column names.

>It perform label - based slicing.

Syntax: For single Column
```python
df.loc[starting_index : ending_index +- 1 : updation, column_name]
```
Syntax: For Multiple Columns
```python
df.loc[starting_index : ending_index +- 1 : updation, [column_name1,column_name2....,column_nameN]]
```

````python
df.columns
````

````python
df.loc[2:5:]
````

````python
df.loc[2::5]
````

````python
df.loc[::5,['name', 'age', 'class', 'gender']]
````

````python
df.loc[2:5:,['name']]
````

## ***iloc[]**

>It an inbuild function / indexer which is used to slice / select rows and columns using their integer index positions, instead of using their label name.

>It perform position - based slicing.

>It work like python lists and numpy array

```python
df.iloc[row_slicing, column_slicing]
```

````python
df.columns
````

````python
df.iloc[2:6,2:6]
````

````python
df.iloc[::5,1:9]
````

````python
df.iloc[::-3,-2::-3]
````

````python
df.iloc[::5,::3]
````

# Boolean Filtering

It is an phenomenon of selecting specific rows from a dataframe based on True / False conditions.

It will return those rows for which condition is True.

Syntax:
```python
df[condition] ----> return rows when condition is true
        OR
df['column_name'][codition] ----> return values from column where condition is True
        OR
df[(condition1) operator (condition2) operator .... operator (conditionN)]
# combine multiple conditions using operator
```

````python
df[df['age']>14]
````

````python
df['class'][df['class'] > 9]
````

````python
df.gender[df['gender'] == 'Male']
````

````python
df[['gender','name','class']][(df['gender'] == 'Male') & (df['class'] > 9)]
````
